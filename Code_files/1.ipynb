```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

# os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")


```

```python
from langchain.agents import create_agent

def get_weather_info(location,temperature):
    """Fetches the latest weather information for a given location."""
    return f"The current weather in {location} is {temperature}."

agent = create_agent(
    model = "google_genai:gemini-3.8-flash",
    tools = [get_weather_info],
    system_prompt = "You are a helpful assistant.",
)
agent

# agent = create_agent(
#     model="groq:openai/gpt-oss-120b",
#     tools=[get_weather_info],
#     system_prompt="You are a helpful assistant."
# )
# agent
```

```python
response = agent.invoke({
   "messages" : [
        {"role":"user","content":"What is the current weather in New York?"}
    ]
})

#response = agent.invoke({"messages":"What is the current weather in New York?"})
```
