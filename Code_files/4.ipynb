```python
import os
from dotenv import load_dotenv
load_dotenv()

api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("openai/gpt-oss-120b",model_provider="groq")
model
```

```python
model.invoke("what is pydantic?")
```

Pydantic library struct the output from AI.

```python
from pydantic import BaseModel,Field

class bollywood(BaseModel):
    name: str = Field(description="Name of the actor")
    age: int = Field(description="Age of the actor")
    movies: list[str] = Field(description="List of movies the actor has been in")
    awards: dict[str,int] = Field(description="Dictionary of awards and their counts")
```

```python
finder = model.with_structured_output(bollywood)
```

```python
details_finder = finder.invoke("Can you give me the details of tej actor mamooty?")
```

```python
details_finder
```
